```python
import pandas as pd
import matplotlib.pyplot as plt

a = pd.read_csv('reported_totals.csv')
a_f = a[a['as_of'] == '2026-02-18'].copy()

mid = a_f['reported_facilities'].median()

a_f['diff'] = a_f['reported_facilities'] - mid

top15 = a_f.sort_values('reported_facilities', ascending=False).head(15)

top15.plot(x='country', y='reported_facilities', kind='bar', figsize=(12, 6))
plt.title('Топ-15 стран по количеству дата-центров')
plt.xlabel('Страна')
plt.ylabel('Количество дата-центров')
plt.show()

top15.plot(x='country', y='diff', kind='bar', figsize=(12, 6), color='pink')
plt.title('Отклонение от медианы по количеству дата-центров')
plt.xlabel('Страна')
plt.ylabel('Отклонение от медианы')
plt.show()
```
